```python
import pandas as pd

fields = [
    "name",
    "license",
    "position",
    "office",
    "cell",
    "work_phone",
    "direct_line",
    "office_phone",
    "email",
    "website",
    "address",
    "profile_url",
    "image_url"
]

records = []

with open(
    "/home/vilas/Datahut-Training/4-09-2026/ewm_data_miner/output/cleaned_data.txt",
    "r",
    encoding="utf-8"
) as file:

    record = {}

    for line in file:
        line = line.strip()

        if not line:
            if record:
                records.append(record)
                record = {}
            continue

        key, value = line.split(":", 1)
        record[key.strip()] = value.strip()

if record:
    records.append(record)

df = pd.DataFrame(records, columns=fields)

df.to_csv(
    "/home/vilas/Datahut-Training/4-09-2026/ewm_data_miner/output/cleaned_data.csv",
    index=False
)

print(f"CSV created successfully: {len(df)} records")
```
