```python
import pandas as pd
```

```python
df=pd.read_csv('sephora_makeup.csv')
df
```

```python
print(df.duplicated().sum())
print(df["url"].duplicated().sum())
```

```python
text_columns = [
    "name",
    "brand",
    "image",
    "category",
    "product_size",
    "shades",
    "description",
    "ingredients",
    "how_to"
]

for column in text_columns:
    df[column] = (
        df[column]
        .fillna("")
        .astype(str)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )
```

```python
df.head()
```

```python
df["price"] = (
    df["price"]
    .str.replace("$", "", regex=False)
    .str.strip()
)
```

```python
df["price"] = pd.to_numeric(
    df["price"],
    errors="coerce"
)
```

```python
print(df["price"].dtype)
print(df["price"].head())
```

```python
print(df["price"].min())
print(df["price"].max())
print((df["price"] <= 0).sum())
```

```python
df["rating"] = pd.to_numeric(
    df["rating"],
    errors="coerce"
)

print(df["rating"].dtype)
```

```python
print("Minimum rating:", df["rating"].min())
print("Maximum rating:", df["rating"].max())
print("Invalid ratings:", ((df["rating"] < 0) | (df["rating"] > 5)).sum())
```

```python
df.dtypes
```

```python
df["reviews"] = pd.to_numeric(
    df["reviews"],
    errors="coerce"
)

print("Reviews dtype:", df["reviews"].dtype)
print("Minimum reviews:", df["reviews"].min())
print("Invalid reviews:", (df["reviews"] < 0).sum())
```

```python
print("Missing URLs:", df["url"].isnull().sum())
print("Invalid URLs:", (~df["url"].str.startswith("https://www.sephora.sg/")).sum())
```

```python
print("Missing categories:", df["category"].isnull().sum())
print("Unique categories:", df["category"].unique())
print("Invalid categories:", (df["category"] != "Makeup").sum())
```

```python
print("Missing product sizes:", (df["product_size"] == "").sum())
print("\nSample product sizes:")
print(df["product_size"].drop_duplicates().head(30).to_string(index=False))
```

```python
invalid_size_values = [
    "Airbrush Flawless Setting Spray",
    "Radiant Creamy Concealer",
    "On Til Dawn Setting Spray",
    "Lustreglass Sheer-Shine Lipstick",
    "Oil Sucker Liquid Blotting Paper Spray",
    "All Nighter Setting Spray"
]

print(df[df["product_size"].isin(invalid_size_values)][
    ["name", "product_size", "shades"]
].to_string(index=False))
```

```python
# Convert empty strings back to NaN
df = df.replace("", pd.NA)

# Remove rows containing any missing value
df = df.dropna().reset_index(drop=True)

print("Dataset shape after removing null rows:", df.shape)
print("\nRemaining missing values:")
print(df.isnull().sum())
```

```python
df.shape
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate URLs:", df["url"].duplicated().sum())
```

```python
print("Unique product sizes:", df["product_size"].nunique())

print("\nProduct sizes:")
print(df["product_size"].drop_duplicates().to_string(index=False))
```

```python
import re

size_pattern = r"^\s*\d+(\.\d+)?\s*(g|kg|ml|l|oz|pcs?|pc)\s*(?:-\s*.+)?\s*$"

invalid_sizes = ~df["product_size"].str.match(
    size_pattern,
    case=False,
    na=False
)

print("Invalid product_size values:", invalid_sizes.sum())
print("Total rows:", len(df))
```

```python
print(
    df.loc[invalid_sizes, ["name", "product_size", "shades"]]
    .to_string(index=False)
)
```

```python
size_from_shades = df["shades"].str.findall(
    r"\b\d+(?:\.\d+)?\s*(?:g|kg|ml|l|oz|pcs?|pc)\b",
    flags=re.IGNORECASE
)

mask = invalid_sizes & size_from_shades.str.len().gt(0)

df.loc[mask, "product_size"] = size_from_shades[mask].apply(
    lambda x: ", ".join(dict.fromkeys(x))
)

print("Product sizes fixed:", mask.sum())
```

```python
print(
    df.loc[invalid_sizes, ["name", "product_size", "shades"]]
    .to_string(index=False)
)
```

```python
invalid_sizes = ~df["product_size"].str.match(
    size_pattern,
    case=False,
    na=False
)

print("Remaining invalid product_size values:", invalid_sizes.sum())
```

```python
df.loc[invalid_sizes, "product_size"] = pd.NA

print("Invalid product_size values remaining:",
      (~df["product_size"].str.match(
          size_pattern,
          case=False,
          na=False
      )).sum())

print("Missing product_size values:",
      df["product_size"].isna().sum())
```

```python
invalid_sizes = (
    df["product_size"].notna()
    & ~df["product_size"].str.match(
        size_pattern,
        case=False,
        na=False
    )
)

print("Invalid product_size values:", invalid_sizes.sum())
print("Missing product_size values:", df["product_size"].isna().sum())
```

```python
print("Missing shades:", df["shades"].isna().sum())
print("Unique shades:", df["shades"].nunique())

print("\nSample shades:")
print(df["shades"].drop_duplicates().head(20).to_string(index=False))
```

```python
shade_size_pattern = r"^\s*\d+(?:\.\d+)?\s*(?:g|kg|ml|l|oz|pcs?|pc)(?:\s*,\s*\d+(?:\.\d+)?\s*(?:g|kg|ml|l|oz|pcs?|pc))*\s*$"

invalid_shades = (
    ~df["shades"].str.match(
        shade_size_pattern,
        case=False,
        na=False
    )
)

print("Values containing only sizes:", (~invalid_shades).sum())
print("Values that are not only sizes:", invalid_shades.sum())
```

```python
size_only_mask = df["shades"].str.match(
    shade_size_pattern,
    case=False,
    na=False
)

df.loc[size_only_mask, "shades"] = pd.NA

print("Size-only shades removed:", size_only_mask.sum())
print("Missing shades after cleaning:", df["shades"].isna().sum())
```

```python
print("Missing descriptions:", df["description"].isna().sum())
print("Empty descriptions:", (df["description"].str.strip() == "").sum())

print("\nSample descriptions:")
print(df["description"].head(5).to_string(index=False))
```

```python
print("Missing ingredients:", df["ingredients"].isna().sum())
print("Empty ingredients:", (df["ingredients"].str.strip() == "").sum())

print("\nSample ingredients:")
print(df["ingredients"].head(5).to_string(index=False))
```

```python
print("Missing how_to:", df["how_to"].isna().sum())
print("Empty how_to:", (df["how_to"].str.strip() == "").sum())

print("\nSample how_to:")
print(df["how_to"].head(5).to_string(index=False))
```

```python
print("========== FINAL VALIDATION ==========")

print("Shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
print("Duplicate URLs:", df["url"].duplicated().sum())

print("\nPrice range:")
print("Min:", df["price"].min())
print("Max:", df["price"].max())

print("\nRating range:")
print("Min:", df["rating"].min())
print("Max:", df["rating"].max())

print("\nReviews:")
print("Min:", df["reviews"].min())
print("Negative reviews:", (df["reviews"] < 0).sum())

print("\nCategories:")
print(df["category"].value_counts())
```

```python
# Remove rows with missing product_size OR shades
df = df.dropna(
    subset=["product_size", "shades"]
).reset_index(drop=True)

print("Final shape:", df.shape)
print("\nMissing values:")
print(df[["product_size", "shades"]].isnull().sum())
```

```python
print("========== FINAL DATASET ==========")
print("Shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
print("Duplicate URLs:", df["url"].duplicated().sum())
```

```python
df.to_csv(
    "sephora_makeup_cleaned.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nSaved: sephora_makeup_cleaned.csv")
print("Final shape:", df.shape)
```
