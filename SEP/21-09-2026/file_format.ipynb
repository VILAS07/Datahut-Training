## HEADER AND COLUMN VALIDATION

```python
import csv

with open("sephora_makeup_cleaned.csv", "r", encoding="utf-8", newline="") as f:
    reader = csv.reader(f)

    header = next(reader)
    expected_columns = len(header)

    incorrect_rows = []

    for row_number, row in enumerate(reader, start=2):
        if len(row) != expected_columns:
            incorrect_rows.append((row_number, len(row)))

print("Expected columns:", expected_columns)
print("Header:", header)
print("Rows with incorrect column count:", len(incorrect_rows))

if incorrect_rows:
    print("Incorrect rows:", incorrect_rows[:10])
else:
    print("✓ All rows have the correct number of columns")
```

## 10 — Validate JSONL

```python
valid_lines = 0
invalid_lines = []

with open("data.jsonl", "r", encoding="utf-8") as f:
    for line_number, line in enumerate(f, start=1):
        try:
            record = json.loads(line)

            if isinstance(record, dict):
                valid_lines += 1
            else:
                invalid_lines.append((line_number, "Not a JSON object"))

        except json.JSONDecodeError as e:
            invalid_lines.append((line_number, str(e)))

print("Total lines:", valid_lines + len(invalid_lines))
print("Valid JSON objects:", valid_lines)
print("Invalid lines:", len(invalid_lines))

if not invalid_lines:
    print("Every line contains a valid JSON object")
else:
    print("Invalid JSONL lines found")
    print(invalid_lines[:10])
```

```python
with open("data.jsonl", "r", encoding="utf-8") as f:
    first_record = json.loads(f.readline())

print("JSONL fields:")
print(list(first_record.keys()))

print("\nOriginal CSV fields:")
print(df.columns.tolist())

if list(first_record.keys()) == df.columns.tolist():
    print("Field names are consistent")
else:
    print("Field names differ")
```

## 14 - TSV VALIDATION

```python
import csv

with open("data.tsv", "r", encoding="utf-8", newline="") as f:
    reader = csv.reader(f, delimiter="\t")

    header = next(reader)
    expected_columns = len(header)

    incorrect_rows = []

    for row_number, row in enumerate(reader, start=2):
        if len(row) != expected_columns:
            incorrect_rows.append((row_number, len(row)))

print("Expected columns:", expected_columns)
print("Rows with incorrect column count:", len(incorrect_rows))

if not incorrect_rows:
    print("TSV structure is valid")
else:
    print("TSV structure issue found")
    print(incorrect_rows[:10])
```

```python
with open("data.tsv", "r", encoding="utf-8") as f:
    first_line = f.readline()

print("Tab separators in header:", first_line.count("\t"))

if "\t" in first_line:
    print("Fields are tab-separated")
else:
    print("Tab separator not found")
```

## 17 - JSON LIST ↔ JSONL CONVERSION

```python
with open("data.json", "r", encoding="utf-8") as f:
    json_list = json.load(f)

with open("data.jsonl", "w", encoding="utf-8") as f:
    for record in json_list:
        f.write(json.dumps(record, ensure_ascii=False) + "\n")

print("JSON List converted to JSONL")
print("Records:", len(json_list))
```

```python
jsonl_records = []

with open("data.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        jsonl_records.append(json.loads(line))

with open("jsonl_to_json.json", "w", encoding="utf-8") as f:
    json.dump(jsonl_records, f, indent=4, ensure_ascii=False)

print("JSONL converted back to JSON List")
print("Records:", len(jsonl_records))
```

```python
print("JSON → JSONL → JSON comparison:",
      json_list == jsonl_records)

if json_list == jsonl_records:
    print("JSON List and JSONL contain identical data")
else:
    print("Differences found")
```
